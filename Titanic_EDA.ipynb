```python
sns.countplot(data=df, x='Sex')

plt.title('Number of Passengers by Sex')
plt.xlabel('Sex')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
sns.countplot(data=df, x='Survived')

plt.title('Passenger Survival Distribution')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.show()
```

```python
sns.countplot(data=df, x='Sex', hue='Survived')

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
sns.countplot(data=df, x='Pclass', hue='Survived')

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
sns.histplot(data=df, x='Age', bins=20, kde=True)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
sns.histplot(data=df, x='Age', hue='Survived', bins=20, kde=True)

plt.title('Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
sns.histplot(data=df, x='Fare', bins=30, kde=True)

plt.title('Fare Distribution of Passengers')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
sns.boxplot(data=df, x='Survived', y='Fare')

plt.title('Fare Distribution by Survival')
plt.xlabel('Survival Status')
plt.ylabel('Fare')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.show()
```

```python
correlation = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.countplot(data=df, x='Embarked')

plt.title('Number of Passengers by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
sns.countplot(data=df, x='Embarked', hue='Survived')

plt.title('Survival by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df[['SibSp', 'Parch', 'FamilySize']].head() 
```

```python
sns.countplot(data=df, x='FamilySize')

plt.title('Family Size Distribution')
plt.xlabel('Family Size')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
sns.countplot(data=df, x='FamilySize', hue='Survived')

plt.title('Survival by Family Size')
plt.xlabel('Family Size')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
survival_by_class = df.groupby('Pclass')['Survived'].mean() * 100

survival_by_class.plot(kind='bar')

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

```python
survival_by_gender = df.groupby('Sex')['Survived'].mean() * 100

survival_by_gender.plot(kind='bar')

plt.title('Survival Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Survival Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

# Titanic Dataset – Exploratory Data Analysis (EDA)

## EDA Summary

The dataset contains 418 passenger records and 12 original variables. The analysis examined passenger demographics, survival patterns, passenger class, age, fare, embarkation port, and family size.

### Key Areas Analyzed

- Dataset structure and dimensions
- Data types and descriptive statistics
- Missing values
- Passenger distribution by gender
- Survival distribution
- Survival by gender
- Survival by passenger class
- Age distribution
- Fare distribution
- Survival by embarkation port
- Family size distribution
- Family size and survival
- Duplicate records
- Correlation between numerical variables
- Overall survival rate
- Survival rate by gender and passenger class

### Important Findings

The overall survival rate in this dataset was approximately 36.36%. The analysis also showed differences in survival across gender and passenger class. Age and fare distributions were examined to identify patterns and potential outliers.

### Data Quality

Missing values were identified mainly in the Age and Cabin variables, while Fare contained only a small number of missing values. No completely duplicated rows were identified.

### Conclusion

The exploratory analysis provided an initial understanding of the Titanic passenger data and revealed relationships between passenger characteristics and survival. These findings can be used as a foundation for further statistical analysis or predictive modelling.
